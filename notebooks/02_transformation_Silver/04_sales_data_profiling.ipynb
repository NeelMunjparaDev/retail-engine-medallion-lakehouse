```sql
SELECT * FROM retail_project.bronze.sales_raw
```

```sql
SELECT
product_name,
COUNT(*) AS record_count
FROM retail_project.bronze.sales_raw
GROUP BY product_name
ORDER BY record_count ASC;
```

```sql
SELECT
cust_id,
COUNT(order_id) AS total_order_per_user
FROM retail_project.silver.sales_cleaned
GROUP by cust_id
ORDER BY cust_id ASC;
```

```sql
SELECT COUNT(DISTINCT cust_id) AS dist_cust_count
FROM retail_project.bronze.sales_raw;
```

```sql
SELECT
    COUNT(DISTINCT order_id) AS unique_orders,
    COUNT(order_id) AS total_rows
FROM retail_project.silver.sales_cleaned
WHERE order_status = 'DELIVERED';
```
