```sql

SELECT 'Bronze' as Layer, count(*) as Count FROM retail_project.bronze.sales_raw
UNION ALL
SELECT 'Silver' as Layer, count(*) as Count FROM retail_project.silver.sales_cleaned
UNION ALL
SELECT 'Gold' as Layer, count(*) as Count FROM retail_project.gold.fact_sales
```

```sql
SELECT row_hash, COUNT(*) 
FROM retail_project.silver.sales_cleaned 
GROUP BY row_hash 
HAVING COUNT(*) > 1 
LIMIT 10;
```

```sql
SELECT 
c.region,
c.city,
p.category,
p.product_name,
d.day_name,
SUM(f.final_price) as total_revenue,
COUNT(f.order_id) as total_orders
FROM retail_project.gold.fact_sales f
JOIN retail_project.gold.dim_customers c ON f.customer_key = c.customer_key
JOIN retail_project.gold.dim_products p ON f.product_key = p.product_key
JOIN retail_project.gold.dim_date d ON f.date_key = d.date_key
GROUP BY 1, 2, 3, 4, 5
ORDER BY total_revenue DESC;
```
