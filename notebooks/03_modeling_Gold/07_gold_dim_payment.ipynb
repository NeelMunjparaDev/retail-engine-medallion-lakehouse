```python
from pyspark.sql import functions as F
```

```python
source_table = "retail_project.silver.sales_cleaned"
target_table = "retail_project.gold.dim_payment"
silver_df = spark.table(source_table)
```

```python
new_payment_data = (
    silver_df.select("payment_mode", "payment_status", "order_status", "found_us_via").distinct()
    .withColumn("payment_key",F.sha2(F.concat_ws("||", "payment_mode", "payment_status", "order_status", "found_us_via"), 256)))

new_payment_data.write.format("delta").mode("overwrite").saveAsTable(target_table)
```
